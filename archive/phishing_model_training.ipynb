# Problem Tanımı

Bu projede web sitelerinin phishing (oltalama) amacı taşıyıp taşımadığı makine öğrenmesi yöntemleri ile tahmin edilmektedir.

Veri setindeki hedef değişken `label` sütunudur.

- `0` → Phishing / Zararlı web sitesi
- `1` → Güvenli web sitesi

Bu nedenle problem bir sınıflandırma problemidir.

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

print("Grafik ayarlari hazir.")
```

Bu grafik, veri setindeki güvenli ve phishing web sitelerinin dağılımını göstermektedir. Label değeri 0 phishing, 1 ise güvenli web sitesini temsil etmektedir.

Bu histogram, URL uzunluklarının veri setindeki dağılımını göstermektedir. Çok uzun URL'ler phishing sitelerde daha sık görülebileceği için model açısından önemli bir değişkendir.

```python
# ==============================
# CANLI UYGULAMAYA UYGUN MODEL
# ==============================

import os
import sys
import joblib
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from xgboost import XGBClassifier

# Dataseti oku
df = pd.read_csv("../data/raw/phishing_dataset.csv")

print("Dataset okundu:", df.shape)

# Canli uygulamada gercekten cikarabildigimiz feature kolonlari
live_feature_columns = [
    "URLLength",
    "DomainLength",
    "IsDomainIP",
    "NoOfSubDomain",
    "IsHTTPS",
    "NoOfiFrame",
    "HasExternalFormSubmit",
    "HasPasswordField",
    "NoOfExternalRef",
    "NoOfJS"
]

# X ve y ayir
X_live = df[live_feature_columns]
y_live = df["label"]

# Train-test split
X_train_live, X_test_live, y_train_live, y_test_live = train_test_split(
    X_live,
    y_live,
    test_size=0.2,
    random_state=42,
    stratify=y_live
)

print("X_train_live:", X_train_live.shape)
print("X_test_live:", X_test_live.shape)

# Model olustur
live_model = XGBClassifier(
    n_estimators=200,
    max_depth=6,
    learning_rate=0.1,
    random_state=42,
    eval_metric="logloss"
)

# Model egit
live_model.fit(X_train_live, y_train_live)

print("Canli uygulama modeli egitildi.")

# Tahmin yap
y_pred_live = live_model.predict(X_test_live)

# Degerlendirme
print("Accuracy:", accuracy_score(y_test_live, y_pred_live))
print("\nClassification Report:")
print(classification_report(y_test_live, y_pred_live))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test_live, y_pred_live))

# Models klasorunu garanti olustur
os.makedirs("../models", exist_ok=True)

# Kaydet
joblib.dump(live_model, "../models/xgboost_live_model.joblib")
joblib.dump(live_feature_columns, "../models/live_feature_columns.joblib")

print("\nCanli model kaydedildi:")
print("../models/xgboost_live_model.joblib")
print("../models/live_feature_columns.joblib")
```

```python
import sys
sys.path.append("../app")

import os
import joblib
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from xgboost import XGBClassifier

from feature_extractor import extract_url_features

# Dataseti oku
df = pd.read_csv("../data/raw/phishing_dataset.csv")

# URL sutunu kontrolu
print(df.columns.tolist())

# URL'lerden yeni feature olustur
url_features = df["URL"].apply(extract_url_features)
url_features_df = pd.DataFrame(url_features.tolist())

# Label ekle
url_features_df["label"] = df["label"]

print(url_features_df.head())
print(url_features_df.shape)

# X ve y
X_url = url_features_df.drop(columns=["label"])
y_url = url_features_df["label"]

# Train test split
X_train_url, X_test_url, y_train_url, y_test_url = train_test_split(
    X_url,
    y_url,
    test_size=0.2,
    random_state=42,
    stratify=y_url
)

# Model
url_model = XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.08,
    random_state=42,
    eval_metric="logloss"
)

url_model.fit(X_train_url, y_train_url)

# Tahmin
y_pred_url = url_model.predict(X_test_url)

print("Accuracy:", accuracy_score(y_test_url, y_pred_url))
print(classification_report(y_test_url, y_pred_url))
print(confusion_matrix(y_test_url, y_pred_url))

# Kaydet
os.makedirs("../models", exist_ok=True)

joblib.dump(url_model, "../models/xgboost_url_model.joblib")
joblib.dump(X_url.columns.tolist(), "../models/url_feature_columns.joblib")

print("URL tabanli model kaydedildi.")
```
